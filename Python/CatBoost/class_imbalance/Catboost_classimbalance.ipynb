```
# Imports
import numpy as np
import pandas as pd
from catboost import CatBoostClassifier
from sklearn.datasets import make_classification
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
```

```
# 1. Generate a synthetic imbalanced dataset (75% majority, 25% minority)
X_raw, y_raw = make_classification(
    n_samples=1000,
    n_features=10,
    weights=[0.75, 0.25],
    random_state=42)

# Split data for train and test
X_train, X_test, y_train, y_test = train_test_split(
    X_raw, y_raw, test_size=0.2, random_state=42
)
```

```
# 2. Calculate inverse class weights dynamically
classes = np.unique(y_train)
weights = len(y_train) / (len(classes) * np.bincount(y_train))
class_weights_dict = dict(zip(classes, weights))
print(f"Calculated Class Weights: {class_weights_dict}")
```

```
# 3. Train CatBoost using class_weights
model = CatBoostClassifier(
    iterations=200,
    learning_rate=0.1,
    class_weights=weights,  # Pass array or list of weights matching class indices
    verbose=0,
)
model.fit(X_train, y_train)
```
